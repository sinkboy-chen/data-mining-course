```python
def determine_modal_value(df, column):
    
    # YOUR CODE HERE
    modal_value = df.mode(axis=0)[column][0]
    
    return modal_value
```

<font size="+1" color="blue">**Write your strategy for handling null values in bathrooms and bedrooms**. Justify briefly your choices.</font>

**Answer:** the strategy for handling missing values in `bathrooms`, `bathrooms_text`, `bedrooms` and `beds` will be as follows.

For columns `bathrooms` and `bathrooms_text`:
* (a) If `bathrooms` is not Null and `bathrooms_text` is Null and `bathrooms`==0.5, we fill `bathrooms_text` with "Half-bath". (b) If `bathrooms` is not Null and `bathrooms_text` is Null and `bathrooms`==1.0, we fill `bathrooms_text` with "1 bath". (c) If `bathrooms` is not Null and `bathrooms_text` is Null and `bathrooms`>1.0, we fill `bathrooms_text` with "{bathrooms} baths". (d) If `bathrooms_text` is not Null and `bathrooms` is Null
and `bathrooms_text` is in ["Half-bath", "Private half-bath", "Shared half-bath"], we fill `bathrooms` with 0.5, else we fill `bathrooms` with the float(N) in start of `bathrooms_text`. (e) If both `bathrooms_text` and `bathrooms` are Null, we fill them with the modal value of the original `bathrooms` columns, which is 1.0 and assign `bathrooms_text` as "1 bath".

For columns `bedrooms` and `beds`:
* (a) If `bedrooms` is not Null and `beds` is Null, we fill `beds` with `bedrooms` number. (b) If `beds` is not Null and `bedrooms` is Null and `beds`<2, we fill `bedrooms` with `beds`. (c) If `beds` is not Null and `bedrooms` is Null and `beds`>=2, we fill `bedrooms` with `beds`-1. (d) If both `beds` and `bedrooms` are Null, we fill them both with 1.0.

<font size="+1" color="blue">**Implement your strategy for handling null values in bathrooms and bedrooms**. </font>

```python
def handle_null_values_bathrooms_bedrooms(df):
    df_clean = df.copy()
    
    # YOUR CODE HERE
    for index, row in df_clean.iterrows():
        # beds and bedrooms part
        if pd.isna(row['beds']) and pd.isna(row['bedrooms']):
            df_clean.at[index, "beds"] = 1
            df_clean.at[index, "bedrooms"] = 1
        elif pd.isna(row['beds']) and not pd.isna(row['bedrooms']):
            df_clean.at[index, "beds"] = row['bedrooms']
        elif not pd.isna(row['beds']) and pd.isna(row['bedrooms']):
            if row['beds'] >= 2:
                df_clean.at[index, "bedrooms"] = row['beds'] - 1
            else:
                df_clean.at[index, "bedrooms"] = row['beds']
                
        # bathrooms and bathrooms_text part
        if pd.isna(row['bathrooms']) and pd.isna(row['bathrooms_text']):
            df_clean.at[index, 'bathrooms'] = 1.0
            df_clean.at[index, 'bathrooms_text'] = "1 bath"
        elif not pd.isna(row['bathrooms']) and pd.isna(row['bathrooms_text']):
            if row['bathrooms']==0.5:
                df_clean.at[index, 'bathrooms_text'] = "Half-bath"
            elif row['bathrooms']==1:
                df_clean.at[index, 'bathrooms_text'] = "1 bath"
            else:
                if row['bathrooms']%1==0:
                    df_clean.at[index, 'bathrooms_text'] = f"{int(row['bathrooms'])} baths"
                else:
                    df_clean.at[index, 'bathrooms_text'] = f"{row['bathrooms']} baths"
                    
        elif pd.isna(row['bathrooms']) and not pd.isna(row['bathrooms_text']):
            if row['bathrooms_text'] in ["Half-bath", "Private half-bath", "Shared half-bath"]:
                df_clean.at[index, 'bathrooms'] = 0.5
            else:
                df_clean.at[index, 'bathrooms'] = float(row['bathrooms_text'].split()[0])
    return df_clean
    
df3 = handle_null_values_bathrooms_bedrooms(df2)

# This should show no NULL values in bathrooms or bedrooms
display(null_value_counts(df3))
```

# 6. Handle null values in other columns <font color="blue">[20%]</font>

<font size="+1" color="blue">**Write your strategy for the remaining columns with null values**. Indicate for each column if you will (a) drop the rows containig null values, (b) replace by a modal value, (c) replace by a zero or empty string. Justify briefly each choice.</font>

**Answer:** the strategy for handling missing values will be as follows.

Columns related to the description of the property:

* Column `description`: replace with an empty string, as an empty description is valid.

Columns related to minimum and maximum number of nights (`*_nights`): 

* Column `minimum_nights`, `maximum_nights`, `minimum_minimum_nights`, `maximum_minimum_nights`, `minimum_maximum_nights`, `maximum_maximum_nights`: drop, since they has very small amount of null data, dropping them won't have significant affect.

Columns related to availability, price, and license:

* Column `has_availability`, `price`, `price_quote_checkin_date`, `price_quote_checkout_date`, `price_quote_total_price`, `price_quote_price_per_night`, `price_quote_raw`, `license`: drop, since we cannot handle listings if we do not know if they are available, how much they cost, or if they are not licensed.

* Column `estimated_revenue_l365d`: will be automatically dropped out if we dropped the null values in `price`

Columns related to reviews:
* Column `review_scores_rating`, `review_scores_accuracy`, `review_scores_cleanliness`, `review_scores_checkin`, `review_scores_location`, `review_scores_value`, `review_scores_communication`: replace with average, since we do not want to penalize listings that have not received a review, so we will replace all scores in the reviews by their average values

* Column `reviews_per_month`: replace with 0, since without the `reviews_per_month` likely indicates there is not reviews yet

* Column `first_review` `last_review`: replace `first_review` and `last_review` with the smallest `first_review` and `last_review`, since without the dates likely indicates that there is no review, and making the `first_review` `last_review` far from current time is more logical

<font size="+1" color="blue">**Implement your strategy for handling the remaining null values**. </font>

```python
def handle_remaining_null_values(df):
    df_clean = df.copy()
    
    # YOUR CODE HERE
    df_clean["description"] = df_clean["description"].fillna("")
    
    df_clean = df_clean.dropna(axis=0, subset=[
        'has_availability',
        'price',
        'license',
        'price_quote_checkin_date',
        'price_quote_checkout_date',
        'price_quote_total_price',
        'price_quote_price_per_night',
        'price_quote_raw']
                               )
    
    df_clean = df_clean.dropna(axis=0, subset=['minimum_nights', 'maximum_nights', 'minimum_minimum_nights', 'maximum_minimum_nights', 'minimum_maximum_nights', 'maximum_maximum_nights'])
    
    replace_with_average_columns = [
        'review_scores_rating',
        'review_scores_accuracy',
        'review_scores_cleanliness',
        'review_scores_checkin',
        'review_scores_location',
        'review_scores_value',
        'review_scores_communication'
        ]
    
    for column in replace_with_average_columns:
        df_clean[column] = df_clean[column].fillna(df_clean[column].mean())
        
    df_clean["reviews_per_month"] = df_clean["reviews_per_month"].fillna(0)
        
    df_clean["first_review"] = df_clean["first_review"].fillna(df_clean["first_review"].min())
    df_clean["last_review"] = df_clean["last_review"].fillna(df_clean["last_review"].min())
    
    return df_clean
    
df4 = handle_remaining_null_values(df3)

# This should show no column with a null value
display(null_value_counts(df4))
```
